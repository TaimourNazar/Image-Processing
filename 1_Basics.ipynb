# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">Imports</span>

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">Read Image</span>

```python
img_path = "/kaggle/input/cameraman/cameraman.png";
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
```

# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">Show Image</span>

```python
def show_image(img):
    plt.figure(figsize=(3,3))
    plt.imshow(img, cmap='gray')
    plt.axis('off')
    plt.show()

show_image(img)
```

# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">1-Resize/Scale</span>

```python
scaled_down = cv2.resize(img, (0,0), fx=0.2, fy=0.2)
scaled_down_interpolated = cv2.resize(img, (0,0), fx=0.2, fy=0.2, interpolation=cv2.INTER_LINEAR)
scaled_up = cv2.resize(img, (0,0), fx=2, fy=2)

show_image(scaled_up)
show_image(scaled_down_interpolated)
show_image(scaled_down)
print(scaled_down)
```

# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">2-Crop</span>

```python
height, width = img.shape

crop_area = img[height//4:3*height//4, width//4:3*width//4]
show_image(crop_area)
```

# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">3-Rotate</span>

```python
center = (width//2, height//2)
rotation_matrix = cv2.getRotationMatrix2D(center, angle=45, scale=1) # rotation matrix
img_rotated = cv2.warpAffine(img, rotation_matrix, img.shape, flags=cv2.INTER_LINEAR) # affine transformation
show_image(img_rotated)
```

# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">4-Flip</span>

```python
flip_horizontal = cv2.flip(img, 1)
flip_vertical = cv2.flip(img, 0)
flip_both = cv2.flip(img, -1)

show_image(flip_horizontal)
```

# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">5-Brightness</span>

```python
increase_brightness = np.clip(img.astype(np.int16)+50, 0, 255).astype(np.uint8)
show_image(increase_brightness)

decrease_brightness = np.clip(img.astype(np.int16)-50, 0, 255).astype(np.uint8)
show_image(decrease_brightness)
```

# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">6-Convert to Binary</span>

```python
_, binary_img = cv2.threshold(img, 128,255, cv2.THRESH_BINARY)
show_image(binary_img)
```

# <span style="display:block; color:#0066cc; font-family: 'Consolas', monospace; border-top:6px solid #ff6b6b; padding:15px; background-color:#f0f8ff; border-radius:8px; font-weight: bold; letter-spacing:1px">7-Histogram of Pixel Intensities</span>

```python
plt.figure(figsize=(8,4))
plt.hist(img.ravel(), bins=256, range=[0,256], color='blue', alpha=0.7)
plt.title("Histogram")
plt.xlabel("Frequency")
plt.ylabel("Pixel Intensities")
plt.grid(True, alpha=0.5)
plt.show()
```
